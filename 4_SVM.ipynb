```python
from sklearn import svm
import numpy as np
import pandas as pd
from sklearn.model_selection import train_test_split, cross_val_score, KFold
from sklearn.model_selection import StratifiedKFold, GridSearchCV, cross_validate
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import mean_squared_error
import matplotlib.pyplot as plt
```

# Data Processing

In this section, we cleaned the data by dropping NA. Also, we dropped column REP, county_name, and State. Because the unit of each column is different, we decide to scale each column. 

```python
train = pd.read_csv('training_data.csv')
train.drop(columns = ['Unnamed: 0'], inplace = True)
train.dropna(inplace = True)
```

```python
train
```

```python
train.columns
```

```python
train_cleaned = train.drop(columns = ['REP', 'county_name', 'State','Registerd'])
```

```python
y = train_cleaned['DEM']
x = train_cleaned.drop(columns = ['DEM'])

scaler = StandardScaler()
x = scaler.fit_transform(x)
```

```python
x
```

# Hyper Parameter Tunning

```python
svm_mod = svm.SVR()
GS_SVM =  GridSearchCV(svm_mod, param_grid = {'C':np.linspace(0.01, 10),'epsilon':np.linspace(0.01, 0.2, 10)}, scoring = 'neg_mean_squared_error', cv = 10)
```

```python
GS_SVM.fit(x,y)
```

```python
print(GS_SVM.best_params_)
print(GS_SVM.best_score_)
```

```python
np.mean(cross_val_score(svm.SVR(C=0.6,epsilon=0.02),x,y,scoring = 'neg_mean_squared_error',cv=10))
```

# Test Data Processing

```python
test = pd.read_csv('Georgia_data.csv')
test.drop(columns = ['Unnamed: 0'], inplace = True)
test.dropna(inplace = True)
```

```python
test_cleaned = test.drop(columns = ['REP', 'county_name', 'State','Registerd'])
y_test = test_cleaned['DEM']
x_test = test_cleaned.drop(columns = ['DEM'])

scaler = StandardScaler()
x_test = scaler.fit_transform(x_test)
```

```python
test_cleaned
```

# Test Best Model

```python
svm_mod = svm.SVR(C=0.6,epsilon=0.02)
svm_mod.fit(x,y)
```

```python
Y_pred=svm_mod.predict(x_test)
np.mean((Y_pred - y_test)**2)
```

```python
Y_pred
```

```python
result = pd.DataFrame()
result['Potential'] = test['Registerd']*Y_pred
result['Real'] = test['Registerd']*y_test
result['county'] = test['county_name']
result['Difference'] = result['Potential'] - result['Real']
#sum(result['Difference'] >= 0)
#result
```
